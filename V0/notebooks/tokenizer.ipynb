```python
# a tokenizer is a tool that breaks down text into smaller pieces called tokens
# these tokens are not necessarily words, they can be punctuation marks, numbers, or other meaningful elements of the text
# tokenization is an important step in creating a language model because 
# language models are a type of nn and they require a way to process the text data to be numerical values
# thats why we need to tokenize the data
# this helps in finding all words first and then we can convert them into numerical values using a technique called word embedding
# tokenization helps in only saving the unique words and splitting text into pieces to make them ready for the embedding layer
# the embedding layer then takes each individuaal token and places it into a higher dimensional vector space
# where similar meaning tokens are placed closer together
# the model learns these relationships from the embedding layer
```
